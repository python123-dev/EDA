# 05 - The `apply()` Function
`apply()` runs **your own function** on every value (or every row) of a column.
Use it when pandas has no ready-made method for what you want.

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "name":  ["Arun", "Bala", "Chitra", "Divya", "Elango"],
    "marks": [78, 45, 88, 72, 91],
    "attendance": [92, 80, 95, 70, 98],
})
df
```

## 1. apply with a normal function

```python
def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "Fail"

df["grade"] = df["marks"].apply(get_grade)
df
```

## 2. apply with a lambda (one-line function)

```python
df["marks_out_of_10"] = df["marks"].apply(lambda m: m / 10)
df
```

```python
df["name_upper"] = df["name"].apply(lambda n: n.upper())
df
```

## 3. apply on rows - use `axis=1`

```python
def check_eligible(row):
    if row["marks"] >= 50 and row["attendance"] >= 75:
        return "Eligible"
    return "Not eligible"

df["exam_status"] = df.apply(check_eligible, axis=1)
df
```

## 4. `map()` - replace values using a dictionary

```python
df["grade_points"] = df["grade"].map({"A": 10, "B": 8, "C": 6, "Fail": 0})
df
```

## Practice on the real data
Tip: for simple things like `marks / 10`, direct arithmetic is faster than `apply`.
Use `apply` when you need `if / else` logic.

```python
students = pd.read_csv("data/students.csv")
students["grade"] = students["marks"].apply(get_grade)
students[["name", "marks", "grade"]].head(8)
```

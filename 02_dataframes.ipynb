# 02 - Pandas DataFrame
A **DataFrame** is a table: rows and columns. Each column is a Series.

```python
import pandas as pd
```

## 1. Create a DataFrame from a dictionary

```python
df = pd.DataFrame({
    "name":  ["Arun", "Bala", "Chitra", "Divya"],
    "dept":  ["CSE", "ECE", "CSE", "IT"],
    "marks": [78, 65, 88, 72],
})
df
```

## 2. Basic information about the table

```python
print("Shape   :", df.shape)       # (rows, columns)
print("Columns :", list(df.columns))
print(df.dtypes)
```

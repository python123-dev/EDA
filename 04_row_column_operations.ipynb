# 04 - Row and Column Operations
Add, rename, delete, sort and combine.

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "name":  ["Arun", "Bala", "Chitra", "Divya"],
    "maths": [78, 65, 88, 72],
    "physics": [70, 60, 90, 68],
})
df
```

## 1. Add a new column

```python
df["total"] = df["maths"] + df["physics"]
df["average"] = df["total"] / 2
df
```

```python
df["result"] = "Pass"                      # same value for all rows
df.loc[df["average"] < 65, "result"] = "Fail"   # change some rows
df
```

## 2. Rename columns

```python
df = df.rename(columns={"maths": "math_marks", "physics": "physics_marks"})
df
```

## 3. Drop (delete) columns

```python
df = df.drop(columns=["result"])
df
```

## 4. Sort rows

```python
df.sort_values("total")
```

```python
df.sort_values("total", ascending=False)
```

## 5. Add a new row

```python
new_row = pd.DataFrame({
    "name": ["Elango"], "math_marks": [55], "physics_marks": [50],
    "total": [105], "average": [52.5],
})
df = pd.concat([df, new_row], ignore_index=True)
df
```

## 6. Drop a row (by index label)

```python
df = df.drop(1)         # removes the row with index 1 (Bala)
df
```

```python
df = df.reset_index(drop=True)    # make the index 0,1,2... again
df
```

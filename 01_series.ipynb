# 01 - Pandas Series
A **Series** is a single column of data with labels (called the *index*).
Think of it as a list that remembers a name for every item.

```python
import pandas as pd
```

## 1. Create a Series from a list

```python
marks = pd.Series([78, 65, 88, 72, 55])
marks
```

The left side (0, 1, 2 ...) is the **index**. The right side is the **values**.

```python
print(marks.index)
print(marks.values)
```

## 2. Custom index and creating from a dictionary

```python
marks = pd.Series([78, 65, 88, 72, 55], index=["Arun", "Bala", "Chitra", "Divya", "Elango"])
marks
```

```python
city_data = {"Arun": "Chennai", "Bala": "Madurai", "Chitra": "Chennai"}
cities = pd.Series(city_data)
cities
```

## 3. Access items

```python
print(marks["Chitra"])        # by label
print(marks.iloc[0])          # by position
marks[["Arun", "Divya"]]      # many labels at once
```

## 4. Maths on a Series (applies to every item)

```python
print(marks + 5)              # grace marks
print(marks / 100)            # as fraction
```

## 5. Quick statistics

```python
print("Mean   :", marks.mean())
print("Max    :", marks.max())
print("Min    :", marks.min())
print("Top    :", marks.idxmax())   # label of the maximum
marks.describe()
```

## 6. Filtering with a condition

```python
passed = marks[marks >= 60]
passed
```

## 7. value_counts - how many times does each value appear?

```python
city = pd.Series(["Chennai", "Madurai", "Chennai", "Trichy", "Chennai", "Madurai"])
city.value_counts()
```

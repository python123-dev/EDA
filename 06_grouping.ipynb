# 06 - Grouping with `groupby()`
Split the data into groups → calculate something for each group → show the result.
Example: *average marks of each department*.

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "name":   ["Arun", "Bala", "Chitra", "Divya", "Elango", "Fathima"],
    "dept":   ["CSE", "ECE", "CSE", "IT", "ECE", "CSE"],
    "gender": ["M", "M", "F", "F", "M", "F"],
    "marks":  [78, 65, 88, 72, 55, 91],
})
df
```

## 1. Group and calculate

```python
df.groupby("dept")["marks"].mean()
```

```python
df.groupby("dept")["marks"].max()
```

```python
df.groupby("dept")["marks"].sum()
```

## 2. Count the members of each group

```python
df.groupby("dept").size()
```

## 3. Many calculations at once with `agg`

```python
df.groupby("dept")["marks"].agg(["count", "mean", "min", "max"])
```

## 4. Group by two columns

```python
df.groupby(["dept", "gender"])["marks"].mean()
```

## 5. Pivot table - the same idea in table form

```python
pd.pivot_table(df, values="marks", index="dept", columns="gender", aggfunc="mean")
```

## Practice on the real data

```python
students = pd.read_csv("data/students.csv")
students.groupby("dept")[["marks", "attendance"]].mean()
```

```python
summary = students.groupby("city")["marks"].agg(["count", "mean"]).round(1)
summary.sort_values("mean", ascending=False)
```

# 03 - Indexing and Selecting Data
How do we pick the columns / rows we want?

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "name":  ["Arun", "Bala", "Chitra", "Divya", "Elango"],
    "dept":  ["CSE", "ECE", "CSE", "IT", "MECH"],
    "marks": [78, 65, 88, 72, 55],
    "city":  ["Chennai", "Madurai", "Chennai", "Coimbatore", "Trichy"],
})
df
```

## 1. Select columns

```python
df["name"]              # one column -> Series
```

```python
df[["name", "marks"]]   # many columns -> DataFrame (note the double brackets)
```

## 2. `iloc` - select by **position** (0, 1, 2 ...)

```python
print(df.iloc[0])           # first row
```

```python
df.iloc[1:3]               # rows 1 and 2 (end is NOT included)
```

```python
df.iloc[0:3, 0:2]          # rows 0-2, columns 0-1
```

## 3. `loc` - select by **label**

```python
df.loc[2]                         # row with index label 2
```

```python
df.loc[0:2, ["name", "marks"]]   # loc INCLUDES the end label
```

## 4. Boolean filtering (conditions)

```python
df[df["marks"] > 70]
```

```python
df[df["dept"] == "CSE"]
```

Combine conditions with `&` (and), `|` (or). Each condition needs its own brackets.

```python
df[(df["dept"] == "CSE") & (df["marks"] > 80)]
```

```python
df[(df["city"] == "Chennai") | (df["marks"] < 60)]
```

```python
df[df["dept"].isin(["CSE", "IT"])]
```

## 5. Use a column as the index

```python
df2 = df.set_index("name")
df2
```

```python
df2.loc["Chitra", "marks"]
```
